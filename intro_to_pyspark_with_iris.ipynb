Sepal length is double (decimal), so we can proceed with numeric filters

```python
# This code creates temporary view of the database called iris_view to be used in SQL query
iris_df.createOrReplaceTempView("iris_view")
```

```python

# Use the LIKE keyword to filter for species that start with "v"
spark.sql("""
SELECT *
FROM iris_view
WHERE species LIKE 'v%'
""").show()
```

```python
# Group by species, find mean sepal width and max sepal length
spark.sql("""
SELECT
    species,
    AVG(sepal_width) AS mean_sepal_width,
    MAX(sepal_length) AS max_sepal_length
FROM iris_view
GROUP BY species
""").show()
```

```python
# Replace species names with initials
from pyspark.sql.functions import when, col
iris_initials_df = iris_df.withColumn("species",
    when(col("species") == "virginica", "VI")
    .when(col("species") == "versicolor", "VE")
    .when(col("species") == "setosa", "SE")
    .otherwise(col("species"))
)
display(iris_initials_df)
```

```python
#Add missing values, then drop rows with no species
irisna = iris_df.replace(0.2, None)

irisna.show()
```

```python
# Replace missing values as efficiently as possible. 
# Because there are many missing values - droppoing rows or replacing with '0' is not a good solutuon. 
# We will replace NULLs with a mean value ofd column 'petal width'.

from pyspark.sql.functions import avg

mean_petal_width = round(
    irisna.select(avg("petal_width")).first()[0],
    1
)

irisna_filled = irisna.fillna(
    {"petal_width": mean_petal_width}
)

irisna_filled.show()
```

```python
# Have to re-activate the table after a break. Databricks times off seessions aftera while  
iris_df = spark.table("iris")
```

```python
# Create two DataFrames 
from pyspark.sql.functions import avg, round

irisavg = iris_df.groupBy("species").agg(
    round(avg("sepal_length"), 1).alias("avg_sepal_length")
)

from pyspark.sql.functions import max

irismax = iris_df.groupBy("species").agg(
    max("sepal_length").alias("max_sepal_length")
)
```

```python
# Join on species
iris_joined = irisavg.join(
    irismax,
    on="species",
    how="inner"
)

iris_joined.show()
```

```python
# Store a Dataframe as an SQL view
iris_joined.createOrReplaceTempView("iris_joined_view")
```

```python
# Run a simple SQL SELECT quesry using PySpark
spark.sql("""
SELECT species,
       avg_sepal_length,
       max_sepal_length
FROM iris_joined_view
""").show()
```
